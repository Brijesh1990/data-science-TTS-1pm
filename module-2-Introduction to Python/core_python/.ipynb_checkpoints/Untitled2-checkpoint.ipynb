```python
# if else : if is executed when condition is true if condition is fasle else is executed
""" syntax 

     if condition:
        statements
     else:
        statements
"""

age=17
if age>17:
    print('i am eligible for voting')
else:
    print('sorry you are not eleigible')
          
```

```python
# w.a.p to chek an year is leap year or not 
# year%4==0 years is leap year

year=int(input('Enter Your year : '))
# placeholder="Enter only digit"
if year%4==0:
    print('Years is Leap year')
else:
    print('Years is not a Leap year')
```

```python
# if elif : if is executed when condition is true elif is executed to check mutiple true condition
""" syntax 

  if condition:
    statements
  elif condition:
     statements
  elif condition:
     statements
  elif condition:
     statements
  else:
    statements
  """

a=int(input('Enter a number :'))
b=int(input('Enter b number :'))
if a>b:
    print('a is greater than b')
elif b>a:
    print('b is greater than a')
else:
    print('a and b both are equal')
    
```
