```python
# w.a.p to print a somple interest using function take input from users
def si():
    p=int(input("Enter a principle ammount:"))
    n=int(input("Enter a number of year :"))
    r=int(input("Enter a ROI :"))
    simple=p*n*r/100
    print("simple interest of principle ammount is :",simple)

si()
    
    
  
```
