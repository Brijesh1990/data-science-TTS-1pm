```python
# # nested if statements 
# nested is used to condition within another condition
""" syntax 

   if condition:
      if condition:
        statements

    else:
      statements

"""

a=10
b=5
if a>b:
    if a!=0 and b!=0:
        print("a is greter than b and both are positive numbers")
else:
    print("a is less than b")
        
```

```python
n1=int(input("Enter n1 values :"))
n2=int(input("Enter n2 values :"))
n3=int(input("Enter n3 values :"))
if n1>n2 and n1>n3:
    print("N1 is max numbers")
elif n2>n3 and n2>n1:
    print("N2 max numbers")
elif n3>n1 and n3>n2:
    print("n3 max numbers")
else:
    print("somthing went wrong while checking max numbers")

    
    
```
