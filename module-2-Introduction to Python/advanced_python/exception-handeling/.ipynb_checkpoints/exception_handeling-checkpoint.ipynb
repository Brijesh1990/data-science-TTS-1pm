```python
# what is exception handeling 
# exception handeling is used to find unexcepted (like invalid input or missing files) error handle 
# basic syntax of exception handeling .. 
# try:
#     statements
# except:
#     statements

n=10;
try:
    res=n/0
except ZeroDivisionError:
    print("Can't  be divided by zero")
    
    
```

```python
# complete exception handeling 
# try:
#     statements
# except:
#     statements
# else:
#     statements
# finally:
#     statements
# #there are four main keyword to handel exception handeling 

# try:try will be excuted block when condition is true
# except:catches and handels the error if one occurs
# else:executes only if no exception occurs in try
# finally: runs if inally result get without exceptions 

n=input("Enter a Name :")
try:
    print("Name is :",n)
except:
    print('error get')
```
