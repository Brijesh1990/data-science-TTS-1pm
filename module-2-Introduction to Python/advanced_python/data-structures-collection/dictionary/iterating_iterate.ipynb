```python
# iterating dictionary with its key and values 
data={1:"devanshu",2:"lotus",3:"banana",4:"apple",5:"pineapple"}
# iterate values over key 
for key in data:
    print(key)
```

```python
# iterating dictionary with its key and values 
data={1:"devanshu",2:"lotus",3:"banana",4:"apple",5:"pineapple"}
# iterate dict over the value 
for key in data.values():
    print(key)
```

```python
# iterating dictionary with its key and values 
data={1:"devanshu",2:"lotus",3:"banana",4:"apple",5:"pineapple"}
# iterate dict over the key and value 
for key,value in data.items():
    print(f"{key}:{value}")
```
