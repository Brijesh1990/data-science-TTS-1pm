```python
# adding data in dictionary
emp={1:"brijesh",2:"devanshu",3:"meet"}
emp[4]="rajesh"
print(emp)
```

```python
# updating data in dictionary
emp={1:"brijesh",2:"devanshu",3:"meet"}
emp[1]="hetvi"
print(emp)
```

```python
# del data in dictionary
emp={1:"brijesh",2:"devanshu",3:"meet"}
del emp[2]
print(emp)
```

```python
# clear data in dictionary
emp={1:"brijesh",2:"devanshu",3:"meet"}
emp.clear()
print(emp)
```

```python
# pop deleted data at the  key from dictionary
emp={1:"brijesh",2:"devanshu",3:"meet"}
emp.pop(2)
print(emp)
```

```python
emp={1:"brijesh",2:"devanshu",3:"meet"}
emp.popitem()
print(emp)
```
