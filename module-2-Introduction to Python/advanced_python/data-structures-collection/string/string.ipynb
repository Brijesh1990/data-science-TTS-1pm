```python
# what is string  ?
# string is  a set of character enclosed '' or "" or ''' '''  or """ """ i.e called 
# string litrals : How to defined a string string literals 
# examples :  '' or "" or ''' '''  or """ """
# string is a immutable (you can changed or append | delete | add | update )


abc="brijesh pandey"
name='hi i am brijesh'
d="""i am brijesh how are you"""
e='''Hey i'am brijesh kumar pandey'''

print(abc + "\n" + name + "\n"+d+"\n"+e)
print(type(abc + "\n" + name + "\n"+d+"\n"+e))
```

```python
# accessing a specific character in string 
name="Brijesh kumar pandey"
print(name[0]) #
print(name[4])
```

```python
# accessing character from end 
name="Brijesh kumar pandey"
print(name[-1]) 
print(name[-4])
```

```python
# string slicing 
name="brijesh how are you"
print(name[1:6])
print(name[2:5])
print(name[-1:])
print(name[::-1])
```
