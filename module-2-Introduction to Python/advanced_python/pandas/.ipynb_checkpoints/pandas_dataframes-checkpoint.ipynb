```python
# geting series from array with its label
import pandas as pd 
res=pd.Series([10,20,30,40,50,60])
print(res)
```

```python
# create a series from dictionary 
import pandas as pd 
res=pd.Series({'name':'Brijesh kumar pandey'})
print(res)
```

```python
import pandas as pd 
res=pd.DataFrame({

       'name':['brijesh','devanshu','rajesh'],
       'Age':[34,28,35], 
       'Salary':[89500,35500,95000]

})
print(res)
```

```python
# from Numpy array 
import numpy as np
import pandas as pd 
arr=np.array([[1,2],[3,4],[5,6]])
res=pd.DataFrame(arr, columns=['x','y'])
print(res)
```

```python
# data selection & indexing 
# column selection 

import pandas as pd 
res=pd.DataFrame({

       'name':['brijesh','devanshu','rajesh'],
       'Age':[34,28,35], 
       'Salary':[89500,35500,95000]

})
# print(res['name'])
print(res['name','Age'])
```
