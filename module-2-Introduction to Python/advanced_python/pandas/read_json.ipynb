```python
# json stands for javascript object notation 
# A pandas used to read a json file content in python 
# A json  contains plain text but has the formate of an object and well known as json formate and pandas is used to read it.
# create a python file to read content of json
# create a csv file (employee.csv o data.json)
# json file is used for web services or api content in .json 
# read a file content of json we used read_json() method in python
import pandas as pd 
res=pd.read_json('employee.json')
print(res)
```
