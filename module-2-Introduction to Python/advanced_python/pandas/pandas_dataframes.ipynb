```python
# data selection & indexing 
# column selection 

import pandas as pd 
res=pd.DataFrame({

       'name':['brijesh','devanshu','rajesh'],
       'Age':[34,28,35], 
       'Salary':[89500,35500,95000]

})
# print(res['name'])
# print(res[['name','Age']])
print(res[['name','Age','Salary']])
```
