```python
# create a mean values and add on mnull places in csv
# mean_value = total_sum / count
import pandas as pd
res=pd.read_csv("data.csv")
res1=res["Experience_Years"].mean()
res.fillna({"Experience_Years":res1}, inplace=True)
print(res.to_string())
```

```python
# create a mode values and add on null places in csv
# mean_value = total_sum / count

import pandas as pd
res=pd.read_csv("data.csv")
res1=res["Experience_Years"].mode()
res.fillna({"Experience_Years":res1}, inplace=True)
print(res.to_string())
```
