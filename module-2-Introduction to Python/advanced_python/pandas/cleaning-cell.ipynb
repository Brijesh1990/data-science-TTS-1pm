```python
# pandas is used to clean cell
# pandas can be used to removed rows 
# pandas used some method to removed rows 
# dropna() : method returns a new datarame and will not change the original
# inplace () : change the original dataframe dropna(implace=true)
import pandas as pd 
res=pd.read_csv('employee.csv')
new_res=res.dropna();
print(new_res.to_string())

```

```python
import pandas as pd 
res=pd.read_csv('employee.csv')
res.dropna(inplace=True);
print(res.to_string())
```

```python
# this will be insert inplace of null in Experience_Years value 10 on every null values 
import pandas as pd 
res=pd.read_csv('employee.csv')
res.fillna({"Experience_Years":21}, inplace=True);
print(res.to_string())
```
